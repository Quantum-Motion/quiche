# 1. Loading the molecular data files into the Hamiltonian

The data to construct the Hamiltonian for the chemical system of interest are stored two files formats:

- `.fcidump`: This is a simple, widely used ASCII text format designed specifically to store quantum chemistry Hamiltonians and molecular integrals (one- and two-electron integrals)

- `.hdf5`: This is a high-performance binary data format used for the efficient, portable storage of massive scientific datasets.


In this tutorial, we will cover how to construct the relevant Hamiltonians from the given data stored in the above two file formats.

## 1.1 Loading `.fcidump` into the second quantisation Hamiltonian

`.fcidump` files contain the constant energy offset, one-body molecular integrals and two-body molecular integrals and other useful quantities to construct the second quantisation hamiltonian.

Below is an example `.fcidump` file for the H2 molecule in `data/hydrogen.fcidump`: 

```text
&FCI NORB= 2,NELEC= 2,MS2= 0,
 ORBSYM=1,1,
 ISYM= 1,
&END
    0.714139282456140   0 0 0 0
   -1.252705257753715   1 1 0 0
   -0.475697705661824   2 2 0 0
    0.674565096014104   1 1 1 1
    0.181266416982487   2 1 2 1
    0.663537594129644   2 2 1 1
    0.697467384335182   2 2 2 2
```

Below, we load the data into `quiche.core.SecondQuantisedHamiltonian` by using the `quiche.io` module.

````python
from pathlib import Path

from quiche.io import fcidump
````

````python
root_dir = Path().cwd()
path = root_dir / "data" / "hydrogen.fcidump"
sq_ham = fcidump.read(path)

print("num_electrons: ", sq_ham.num_electrons)
print("core_energy: ", sq_ham.core_energy)
print("one body: ", sq_ham.one_body)
print("two body: ", sq_ham.two_body)
````

The `.fcidump` file is loaded into `SecondQuantisedHamiltonian`, and we printed the number of electrons, core energy, one-body and two-body integrals successfully.

## 1.2. Loading `.hdf5` into the qubit Hamiltonian.

The `.hdf5` file contains the qubit Hamiltonians in the byte string format for different fermion-to-qubit mappings with different number of qubits (corresponding to different number of spin orbitals).

Below, we load the data into `quiche.core.ElectronicHamiltonian` by using the `quiche.io` module.

````python
from pathlib import Path

from quiche.io import hamlib
````

````python
path = root_dir / "data" / "H2.hdf5"
# Jordan-Wigner mapping with 4 spin-orbitals (1s for each Hyrdogen; 2 spatial orbitals)
raw_data = hamlib.read_dataset(path, "ham_JW-4")
paulis = hamlib.parse(raw_data)

print("Number of qubits: ", paulis.num_qubits)
print("Number of terms in the qubit Hamiltonian: ", paulis.num_words)
print("1-norm of the qubit Hamiltonian: ", paulis.lam)
````

The `.hdf5` file is loaded into `ElectronicHamiltonian`, and we printed the number of qubits, the number of Pauli terms, and the 1-norm of the qubit Hamiltonian successfully.
